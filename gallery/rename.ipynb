```python
import os
# os.chdir(r"C:\Users\Ahmad\OneDrive\Ahmad - Personal\Desktop\PhD\2026\AD_Wev\gallery")
# import os

def rename_images(prefix="IMG_", start_index=1):
    # Use current working directory (for Jupyter/IPython)
    folder_path = os.getcwd()

    # List all files in the folder
    files = os.listdir(folder_path)

    # Filter only image files
    image_extensions = {".jpg", ".jpeg", ".png", ".gif", ".bmp", ".tiff"}
    image_files = [f for f in files if os.path.splitext(f)[1].lower() in image_extensions]

    # Sort for consistency
    image_files.sort()

    # Rename images sequentially
    for i, filename in enumerate(image_files, start=start_index):
        old_path = os.path.join(folder_path, filename)
        extension = os.path.splitext(filename)[1].lower()
        new_name = f"{prefix}{i}{extension}"
        new_path = os.path.join(folder_path, new_name)

        os.rename(old_path, new_path)
        print(f"Renamed: {filename} → {new_name}")

# Example usage in Jupyter
rename_images(prefix="Photo_", start_index=1)
```
